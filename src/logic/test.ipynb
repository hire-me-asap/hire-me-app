```python
## Create a thread

thread_enpoint = f'{AZURE_OPENAI_ENDPOINT}openai/threads?api-version=2024-05-01-preview'
thread_header = {
  'api-key': AZURE_OPENAI_API_KEY,
  'Content-Type': 'application/json'
}

thread_response = requests.post(thread_enpoint, headers=thread_header) 
thread_response_json = thread_response.json()
THREAD_ID = thread_response_json['id']
# print(THREAD_ID)

USER_QUESTION_ENDPOINT = f"{AZURE_OPENAI_ENDPOINT}openai/threads/{THREAD_ID}/messages?api-version=2024-05-01-preview"

USER_QUESTION_HEADER = {
  'api-key': AZURE_OPENAI_API_KEY,
  'Content-Type': 'application/json'
}

USER_QUESTION_BODY = {
  "role": "user",
  "content": "프론트엔드 개발자가 되고 싶어. 커리어로드맵 그려줘."
}

user_qusetion = requests.post(USER_QUESTION_ENDPOINT, headers=USER_QUESTION_HEADER, json=USER_QUESTION_BODY) 
# print(user_qusetion.status_code)

USER_QUESTION_RUN_ENDPOINT = f"{AZURE_OPENAI_ENDPOINT}openai/threads/{THREAD_ID}/runs?api-version=2024-05-01-preview"
USER_QUESTION_RUN_BODY = {
  "assistant_id": 'asst_PEVAovZx7qtiPZaSE5vhegTq',
}

run_id = requests.post(USER_QUESTION_RUN_ENDPOINT, headers=USER_QUESTION_HEADER, json=USER_QUESTION_RUN_BODY) 
run_id_json = run_id.json()
RUN_ID = run_id_json['id']
# print(RUN_ID)

get_thread_endpoint = f"{AZURE_OPENAI_ENDPOINT}openai/threads/{THREAD_ID}/runs/{RUN_ID}?api-version=2024-05-01-preview"

get_thread_header = {
  'api-key': AZURE_OPENAI_API_KEY
}

get_thread_response = requests.get(get_thread_endpoint, headers=get_thread_header)
# print(get_thread_response.status_code)
# print(get_thread_response.json())
get_thread_json = get_thread_response.json()
thread_id_get = get_thread_json['thread_id']
```

```python
last_endpoint = f"{AZURE_OPENAI_ENDPOINT}openai/threads/{thread_id_get}/messages?api-version=2024-05-01-preview"

last_header = {
  'api-key': AZURE_OPENAI_API_KEY,
  'Content-Type': 'application/json'
}

last_response = requests.get(last_endpoint, headers=last_header)
# print(last_response.status_code)

last_response_json = last_response.json()

assistant_messages = [
    msg for msg in last_response_json['data'] 
    if msg['role'] == 'assistant'
]

if assistant_messages:
    text_value = assistant_messages[0]['content'][0]['text']['value']
    
    # ---json start--- 기준으로 분리
    split_marker = "---json start---"
    if split_marker in text_value:
        text_part, json_part = text_value.split(split_marker, 1)
        text_value = text_part.strip()

        import json
        json_start = json_part.find('{')
        json_end = json_part.rfind('}') + 1
        json_string = json_part[json_start:json_end]

        try:
            text_json = json.loads(json_string)
        except json.JSONDecodeError as e:
            print("❌ JSON 파싱 오류:", e)
            text_json = None
    else:
        print("❗ '---json start---' 구분자가 포함되어 있지 않아요.")
        text_json = None

    print("📄 text_value:\n", text_value)
    print("🧩 text_json:\n", text_json)

else:
    print("Assistant 메시지를 찾을 수 없습니다.")

```

```python
from graphviz import Digraph
from PIL import Image
import io
import json

FONT = 'NeoDunggeunmo Pro'

data = text_json
nodes = data["nodes"]

# Graphviz 객체 생성
dot = Digraph(comment=data.get("title", "Career Roadmap"), encoding='utf-8')
dot.attr(rankdir='LR')
dot.attr('node', fontname=FONT)
dot.attr('edge', fontname=FONT)

# 노드 추가
for node in nodes:
    label = f"{node['name']}\n" + "\n".join(node["details"])
    dot.node(node["name"], label)

# 엣지 추가
for node in nodes:
    for conn in node.get("connections", []):
        dot.edge(node["name"], conn)

# 이미지로 저장
try:
    png_bytes = dot.pipe(format='png')
    image = Image.open(io.BytesIO(png_bytes))
    image.save("career_roadmap.png")
    print("✅ 이미지 저장 완료: career_roadmap.png")
except Exception as e:
    print(f"⚠️ 그래프 이미지 생성 오류: {str(e)}")
    print(f"Dot source:\n{dot.source}")
```
