```python
import gradio as gr
 
# 초기 대화 히스토리 설정
history = [
    {"role": "assistant", "content": "어서오세요! Hire Me ASAP 입니다.\n취업 분석을 원하시면 이력서를 업로드 해보세요.😀"},
    {"role": "user", "content": "안녕 나는 평생 직장을 찾고 있어"}
]
 
# Gradio 테마 설정
theme = gr.themes.Soft(
    primary_hue="gray",
    secondary_hue="stone",
    neutral_hue="zinc",
    font=[
        gr.themes.GoogleFont('Noto Sans Korean '),
        gr.themes.GoogleFont('42dot Sans'),
        gr.themes.GoogleFont('Nanum Gothic '),
        'sans-serif'
    ],
).set(button_transform_hover='*button_primary_background_fill')
 
# 커스텀 CSS 정의
custom_css = """
* {
    font-family: 'NeoDunggeunmo', sans-serif !important;
    font-size: 16px !important;
}
 
/* 공통 버튼 */
.custom-button {
    margin: 10px;
    padding: 5px 5px;
    height: 40px;
    width: 40px;
    display: inline-flex;
    align-items: center;
    justify-content: center;
}
 
/* 업로드 및 체크박스 input 영역 */
.custom-input {
    align-items: top;
}
 
/* 체크박스 스타일 */
input[type="checkbox"] {
    width: 18px;
    height: 18px;
    border: 1px solid #444 !important;
    border-radius: 4px;
    appearance: none;
    background-color: white;
    cursor: pointer;
    position: relative;
}
 
/* 체크된 상태일 때 */
input[type="checkbox"]:checked::before {
    position: absolute;
    top: -2px;
    left: 2px;
    font-size: 16px;
    color: #0d0d0d;
}
 
/* 예시 버튼 배치 - 중앙 정렬 */
example-row {
    position: absolute;
    top: 50%;
    left: 50%;
    transform: translate(-50%, -50%);
    display: flex;
    flex-wrap: wrap;
    justify-content: center;
    gap: 30px;
    width: 100%;
    max-width: 800px;
    padding: 15px;
    z-index: 2;
}
 
/* 예시 버튼 스타일 */
.example-btn {
    background-color: #f2f2f2 !important;
    color: #000;
    padding: 14px 24px;
    border-radius: 12px;
    border: 2px solid #f2f2f2;
    font-size: 0.9rem;
    text-align: center;
    transition: all 0.3s ease;
    box-shadow: 2px 2px 6px rgba(0,0,0,0.1);
    width: 300px;
    min-width: 280px;
    max-width: 310px;
    white-space: nowrap;
}
 
/* 예시 버튼 호버 효과 */
.example-btn:hover {
    background-color: #1a1a1a;
    transform: translateY(-3px);
    cursor: pointer;
}
 
/* 아바타 이미지 크기 조절 */
.gr-avatar {
    width: 40px !important;
    height: 40px !important;
    border-radius: 50%;
}
 
/* 기본 푸터 감추기 */
footer {
    visibility: hidden;
}
 
/* 사이드바 푸터용 flex 컨테이너 */
.sidebar-container {
    display: flex;
    flex-direction: column;
    height: 100%;
}
 
/* 사이드바 하단 고정 푸터 */
.sidebar-footer {
    margin-top: auto;
    font-size: 11px;
    color: #d5d5d5 !important;
    text-align: left;
    line-height: 1.6;
    padding: 12px 0;
    margin-top: 30px;
}
 
/* 중앙 하단 고정 푸터 */
.footer-copyright {
    text-align: center;
    font-size: 9px;
    color: #e3e3e3 !important;
    margin-top: 10px;
    margin-bottom: 5px;
}
"""
 
# 사용자 입력 처리 함수
def handle_user_message(user_msg, chat_history):
    chat_history = chat_history or []
    chat_history.append({"role": "user", "content": user_msg})
    assistant_response = "좋은 질문이에요! 🤖 지금 분석 중입니다."
    chat_history.append({"role": "assistant", "content": assistant_response})
    return chat_history, ""
 
# Gradio 앱 구성
with gr.Blocks(theme=theme, css=custom_css) as demo:
    # 사이드바
    with gr.Sidebar(open=False):
        gr.Markdown("프로필")
        gr.Image(value="profile.png", width="200px", show_label=False, show_download_button=False, show_fullscreen_button=False, show_share_button=False)
        gr.Markdown("🔮 무엇이 궁금하신가요?")
 
        # 사이드바 버튼 및 프롬프트
        sidebar_prompts = [
            ("🔍 인생 회사 찾기", "인생 회사 찾기"),
            ("🎯 나에게 맞는 직무는?", "나에게 맞는 직무는?"),
            ("🔥 내가 보완할 부분은?", "내가 보완할 부분은?"),
            ("📍 커리어 로드맵 그리기", "커리어 로드맵 그리기")
        ]
        sidebar_buttons = [gr.Button(label, size="sm") for label, _ in sidebar_prompts]
 
        gr.Button("📂 대화 내용 저장하기", variant="huggingface", size="sm")
 
        # 소개 및 푸터
        gr.Markdown("🐚 ABOUT US")
        gr.Markdown("서비스 소개 @@@@")
        with gr.Column(elem_classes=["sidebar-footer"]):
            gr.Markdown("""
<div class="sidebar-footer">
                © 2025 hire me ASAP Inc.<br>
                ✉️ contact@hiremeasap.com<br>
                💾 <a href="https://github.com/hire-me-asap" target="_blank" style="color:#d5d5d5;">github.com/hire-me-asap</a><br>
                🔒 <a href="#" style="color:#d5d5d5;">Privacy Policy</a>
</div>
            """)
 
    # 채팅 영역
    # 채팅창
    chatbot = gr.Chatbot(
        height="75vh",
        type='messages',
        container=False,
        avatar_images=("chicken.png", "magicball.jpeg")
    )
    gr.Checkbox(label="🪄 관련 강의 추천 받기", elem_classes=["custom-checkbox"])

    # 입력창 + 업로드 + 전송버튼 (고정)
    with gr.Row():
        gr.UploadButton(label="upload", icon="img.png", file_types=["image"], scale=1, elem_classes=["custom-button"])
        user_input = gr.Textbox(placeholder="무엇이든 물어보세요", show_label=False, scale=30)
        submit_btn = gr.Button("🚀", scale=1, size="lg", elem_classes=["custom-button"])

    # 프롬프트 예시 버튼 (아래쪽에 위치)
    with gr.Row(elem_classes="example-row") as example_container:
        example_texts = [
            "취업할만한 회사를 찾고 싶어",
            "내 커리어 로드맵을 그리고 싶어",
            "내가 개선할 수 있는 부분을 알려줘",
            "나에게 잘 맞는 직무를 찾고 싶어"
        ]
        example_buttons = [gr.Button(text, elem_classes="example-btn") for text in example_texts]
        
    # 예시 버튼 클릭 시 응답 처리
    def handle_example_click(text, chat_history):
        chat_history = chat_history or []
        chat_history.append({"role": "user", "content": text})
        chat_history.append({"role": "assistant", "content": "좋은 질문이에요! 🤖 지금 분석 중입니다."})
        return chat_history, "", gr.update(visible=False)
 
    # 예시 버튼 이벤트
    for btn, text in zip(example_buttons, example_texts):
        btn.click(fn=handle_example_click, inputs=[gr.State(text), chatbot], outputs=[chatbot, user_input, example_container])

    # 사용자 입력 제출
    submit_btn.click(fn=handle_user_message, inputs=[user_input, chatbot], outputs=[chatbot, user_input])
 
    # 예시 버튼 바인딩
    for btn, text in zip(example_buttons, example_texts):
        btn.click(
            fn=handle_example_click,
            inputs=[gr.State(text), chatbot],
            outputs=[chatbot, user_input, example_container]
        )
 
    # 사이드바 버튼 바인딩
    for btn, (_, prompt) in zip(sidebar_buttons, sidebar_prompts):
        btn.click(fn=handle_user_message, inputs=[gr.State(prompt), chatbot], outputs=[chatbot, user_input])
 
# 앱 실행
demo.launch(share=True)
```
